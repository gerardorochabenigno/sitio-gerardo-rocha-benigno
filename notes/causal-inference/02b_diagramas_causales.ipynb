# Diagramas Causales

Un diagrama causal es una herramienta gráfica que permite distinguir entre asociación y causalidad. Para esto, se hace uso de grafos acíclicos dirigidos (Directed Acyclic Graphs).



**Grafo**: Un grafo es una función $G=G(V,A)$ que consiste de nodos $V$ y arcos $A$. Los grafos pueden ser dirigidos o no dirigidos.

**Caminos, ancestros y descendientes**:

- **Camino**: Una sucesión de nodos $A_0, A_1, \ldots, A_n$ donde cada par consecutivo $(A_{k-1}, A_k)$, $k = 1,\ldots,n$, está conectado por un arco en el grafo. Se dice que el camino va de $A_0$ a $A_n$.

- **Camino dirigido**: Un camino en el que todos los arcos siguen la misma dirección, es decir, de $A_0 \rightarrow A_1 \rightarrow \cdots \rightarrow A_n$.

- **Ancestros**: En un grafo dirigido, los ancestros de un nodo $B$ son todos los nodos $A$ desde los cuales existe un camino dirigido hacia $B$.

- **Descendientes**: Los descendientes de un nodo $A$ son todos los nodos $B$ a los que se puede llegar siguiendo caminos dirigidos desde $A$.

- **Grafo acíclico**: Un grafo dirigido en el que no existen caminos dirigidos que regresen a un nodo ya visitado, es decir, no hay ciclos.

  ## Modelación de dependencias

  Consideremos una colección de variables aleatorias $\{X_1,X_2, ...,X_p\}$m la distribución conjunta puede escribirse como:

$$P(X_1, ..., X_p)= P(X_1)P(X_2|X_1),...,P(X_p|X_1,...,X_{p-1})=\prod_{j=1}^p P(X_j|X_1,...,X_{j-1})$$

Donde $X_0 = \empty$ y $(P(\empty)=1)$

Esta idea puede ser unida con la teoría de grafos: Las variables $(X_1,...,X_{j-1})$ son los antecesores de $X_j$. Adicionalmente, podemos suponer que:

$$P(X_j|X_1,...,X_{j-1}) = P(X_j|pa_j)$$

Donde $pa_j$ es un grupo de predecesores mínimo de $X_j$, los cuáles de definen como *padres* de $X_j$. Para $X_1$, $pa_1=\empty$ y podemos reescribir la distribución conjunta de la siguiente forma:

$$P(X_1, ..., X_{j-1})=P(X_j|pa_j)$$

En donde esta última ecuación puede incluir independencias condicionales no especificadas en la ecuación previa.
